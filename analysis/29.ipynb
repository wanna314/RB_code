```python
import glob

all_pulses = glob.glob("data/29/rabi_*cycles_*.npz")

all_data_numbers = {} #dictionary of pulse times: list of data numbers

for sweep in all_pulses:
    n_cycles = int(sweep.split("rabi_")[-1].split("_cycles")[0])
    pulse_times = n_cycles / 590e3
    dnum = int(sweep.split("cycles_")[1].split(".")[0])
    if pulse_times not in all_data_numbers:
        all_data_numbers[pulse_times] = []
    all_data_numbers[pulse_times].append(sweep)

def determine_saturation(dnum):
    data = load_waveforms(f"{dnum}")
    t, V1, V2, V4 = data[1][0], data[1][1], data[2][1], data[4][1]
    N = 200
    t = np.convolve(t, np.ones(N)/N, mode='valid')
    V1 = np.convolve(V1, np.ones(N)/N, mode='valid')
    V2 = np.convolve(V2, np.ones(N)/N, mode='valid')
    t4 = np.convolve(data[4][0], np.ones(10)/10, mode='valid')
    V4 = np.convolve(V4, np.ones(10)/10, mode='valid')

    fig0, ax0 = plt.subplots(figsize=(8, 6))
    # ax0.plot(t, V1, label='Channel 1: Gaussmeter', color='goldenrod')
    ax0.plot(t4, V4*1e-1, label='Channel 4: AWG', color='tab:blue')
    ax0.plot(t, V2*5-2, label='Channel 2: Picoammeter', color='tab:green')
    ax0.set_xlabel('Time (s)')
    ax0.set_ylabel('Scope Signal (arb. units)')
    ax0.set_title(f'Saturation for Data Number {dnum}') 
    ax0.legend()

all_data_numbers
```
